```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python

url = "https://raw.githubusercontent.com/ApnaClassroom/Dataset/main/Students%20Performance.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
```

```python

df = df.drop_duplicates()

print("Duplicate records removed successfully!")
```

```python

print(df.dtypes)
```

```python

print("Math score:")
print(df["math score"].describe())

print("Reading score:")
print(df["reading score"].describe())

print("Writing score:")
print(df["writing score"].describe())
```

```python

print("Total rows:", df.shape[0])
print("Total columns:", df.shape[1])

print("Missing values:")
print(df.isnull().sum().sum())

print("Duplicate records:")
print(df.duplicated().sum())

print("Data cleaning completed!")
```

```python

df.to_csv("Cleaned_Student_Data.csv", index=False)

print("Cleaned dataset saved successfully!")
```

```python

average_marks = df[
    ["math score", "reading score", "writing score"]
].mean()

plt.figure(figsize=(8,5))

average_marks.plot(kind="bar", color=["blue", "green", "orange"])

plt.title("Average Marks by Subject")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.xticks(rotation=0)

plt.show()
```

```python

plt.figure(figsize=(8,5))

plt.hist(df["math score"], bins=10, edgecolor="black")

plt.title("Distribution of Mathematics Marks")
plt.xlabel("Math Marks")
plt.ylabel("Number of Students")

plt.show()
```

```python

plt.figure(figsize=(8,5))

sns.boxplot(
    data=df[["math score", "reading score", "writing score"]]
)

plt.title("Student Marks Comparison")
plt.xlabel("Subjects")
plt.ylabel("Marks")

plt.show()
```

```python

correlation = df[
    ["math score", "reading score", "writing score"]
].corr()

plt.figure(figsize=(7,5))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Student Marks Correlation Heatmap")

plt.show()
```
