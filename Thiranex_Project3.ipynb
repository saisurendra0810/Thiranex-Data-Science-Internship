```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python

url = "https://raw.githubusercontent.com/ApnaClassroom/Dataset/main/Students%20Performance.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")

df.head()
```

```python

print("Dataset Shape:", df.shape)
```

```python

print(df.columns)
```

```python

df.info()
```

```python

df.describe()
```

```python

print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())
```

```python

print("Average Math Marks:", df["math score"].mean())

print("Average Reading Marks:", df["reading score"].mean())

print("Average Writing Marks:", df["writing score"].mean())
```

```python

print("Highest Math Marks:", df["math score"].max())

print("Lowest Math Marks:", df["math score"].min())

print("Highest Reading Marks:", df["reading score"].max())

print("Lowest Reading Marks:", df["reading score"].min())

print("Highest Writing Marks:", df["writing score"].max())

print("Lowest Writing Marks:", df["writing score"].min())
```

```python

average = df[
    ["math score", "reading score", "writing score"]
].mean()

average.plot(kind="bar", figsize=(8,5))

plt.title("Average Marks by Subject")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")

plt.show()
```

```python

plt.figure(figsize=(8,5))

plt.hist(df["math score"], bins=10, edgecolor="black")

plt.title("Math Marks Distribution")
plt.xlabel("Math Marks")
plt.ylabel("Number of Students")

plt.show()
```

```python

plt.figure(figsize=(8,5))

sns.boxplot(
    x="test preparation course",
    y="math score",
    data=df
)

plt.title("Test Preparation vs Math Scores")
plt.xlabel("Test Preparation")
plt.ylabel("Math Marks")

plt.show()
```

```python

correlation = df[
    ["math score", "reading score", "writing score"]
].corr()

plt.figure(figsize=(7,5))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm"
)

plt.title("Student Performance Correlation")

plt.show()
```

```python

high_scores = df[df["math score"] > 80]

print("Students scoring above 80:")
print(len(high_scores))
```
